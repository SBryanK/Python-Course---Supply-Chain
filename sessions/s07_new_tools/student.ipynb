## SQL

SQLite lives inside Python, so there is nothing to install.

```python
import sqlite3

columns_for_sql = ["invoice", "stock_code", "description", "quantity", "price", "country", "revenue"]
sql_table = sales_clean[columns_for_sql]

connection = sqlite3.connect(":memory:")
sql_table.to_sql("sales", connection, index=False)
```

**Predict.** Read the query first. What will it return: how many rows, which columns? Then run it.

```python
# My guess:
query = """
SELECT country, SUM(revenue) AS total_revenue
FROM sales
GROUP BY country
ORDER BY total_revenue DESC
LIMIT 5
"""
pd.read_sql_query(query, connection)
```
